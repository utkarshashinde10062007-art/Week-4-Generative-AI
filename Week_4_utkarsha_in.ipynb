Day 5

```python


from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Machine learning is a part of artificial intelligence.",
    "Python is a popular programming language.",
    "Generative AI can generate text and images.",
    "Deep learning uses neural networks."
]

query = "What is Generative AI?"

vectorizer = TfidfVectorizer()

document_vectors = vectorizer.fit_transform(documents)
query_vector = vectorizer.transform([query])

similarity = cosine_similarity(query_vector, document_vectors)

print("Query:")
print(query)

print("\nSimilarity Scores:")
print(similarity[0])

best_index = similarity.argmax()

print("\nMost Relevant Document:")
print(documents[best_index])
```

```python


def ai_assistant(question):
    if question == "ai":
        return "AI stands for Artificial Intelligence."

    elif question == "ml":
        return "Machine Learning is a branch of Artificial Intelligence."

    elif question == "llm":
        return "LLM stands for Large Language Model."

    elif question == "generative ai":
        return "Generative AI can create new content such as text, images and code."

    else:
        return "Please ask about AI, ML, LLM or Generative AI."



question = "generative ai"


answer = ai_assistant(question)


print("================================")
print("   GENERATIVE AI ASSISTANT")
print("================================")
print("Question:", question)
print("Answer:", answer)
print("================================")
```

Day 7

Objective: To build a simple AI Document Question-Answering Assistant.
Learning: Learned document chunks, embeddings, similarity search and retrieving relevant information.

```python




from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity






document = """
Generative AI is a type of artificial intelligence that can create new content.
Large Language Models are used to understand and generate human-like text.
Prompt engineering helps developers improve AI responses.
Embeddings convert text into numerical vectors.
Retrieval-Augmented Generation retrieves relevant information before generating an answer.
"""






chunks = [
    "Generative AI is a type of artificial intelligence that can create new content.",
    "Large Language Models are used to understand and generate human-like text.",
    "Prompt engineering helps developers improve AI responses.",
    "Embeddings convert text into numerical vectors.",
    "Retrieval-Augmented Generation retrieves relevant information before generating an answer."
]





vectorizer = TfidfVectorizer()

chunk_vectors = vectorizer.fit_transform(chunks)




question = "What is Generative AI?"



question_vector = vectorizer.transform([question])



similarity_scores = cosine_similarity(
    question_vector,
    chunk_vectors
)

best_index = similarity_scores.argmax()

relevant_information = chunks[best_index]




print("======================================")
print(" AI DOCUMENT QUESTION-ANSWERING")
print("======================================")

print("\nQuestion:")
print(question)

print("\nRelevant Information:")
print(relevant_information)

print("\nAnswer:")
print(relevant_information)

print("\nSimilarity Score:")
print(similarity_scores[0][best_index])

print("======================================")
```
